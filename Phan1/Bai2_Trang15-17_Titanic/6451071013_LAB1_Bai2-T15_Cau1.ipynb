**Phân loại thuộc tính:**

| Thuộc tính | Loại |

| name, ticket, cabin, boat, home.dest | Nominal |

| embarked | Nominal |
| age, fare | Quantitative – ratio-scaled |
| sibsp, parch | Quantitative – ratio-scaled (đếm số người) |
| body | Nominal (mã số thi thể, chỉ là định danh) |

**Nhận xét:**

- Dữ liệu có 1.309 hành khách và 14 thuộc tính (4 kiểu nguyên, 3 kiểu thực, 7 kiểu chuỗi).

- Phần lớn khách là hạng 3 (trung vị pclass = 3), là nam (843/1.309) và lên tàu ở cổng S (914/1.307).
- `ticket` có 929 giá trị khác nhau, `name` gần như duy nhất cho mỗi người, nên hai thuộc tính này hầu như không mang thông tin chung để khai thác.
