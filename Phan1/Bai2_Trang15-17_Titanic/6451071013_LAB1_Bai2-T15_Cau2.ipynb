### Chuẩn bị: thực hiện lại các bước của các câu trước

```python
df = pd.read_csv(DATA + 'titanic.csv', dtype={'age': np.float64, 'sibsp': np.int64, 'parch': np.int64})
```

---
## Thực hiện yêu cầu

```python
print(SV)
total = df.isnull().sum().sort_values(ascending=False)
percent = (df.isnull().sum() / df.isnull().count() * 100).round(1).sort_values(ascending=False)
missing_data = pd.concat([total, percent], axis=1, keys=['Total', '%'])
missing_data.head(5)
```

**Nhận xét:**

- 5 thuộc tính thiếu nhiều nhất là `body` (90,8%), `cabin` (77,5%), `boat` (62,9%), `home.dest` (43,1%) và `age` (20,1%).
- `body` và `boat` thực chất là thông tin *sau* thảm họa (số hiệu thi thể, xuồng cứu sinh): dùng chúng để dự đoán sống sót là "biết trước đáp án", nên cần bỏ.
- `cabin` thiếu nhiều nhưng ký tự đầu (boong) có thể hữu ích; `age` chỉ thiếu 20% và rất quan trọng, nên cần điền giá trị thay vì xóa.
