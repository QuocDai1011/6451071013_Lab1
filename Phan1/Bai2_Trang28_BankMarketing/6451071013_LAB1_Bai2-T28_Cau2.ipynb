### Chuẩn bị: thực hiện lại các bước của các câu trước

```python
df = pd.read_csv(DATA + 'bank-full.csv', sep=';')
df['y_num'] = (df['y'] == 'yes').astype(int)   # dùng để tính tỷ lệ đăng ký
```

---
## Thực hiện yêu cầu

```python
print(SV)
df.isnull().sum().sum()   # không có NaN thực sự
```

Bộ dữ liệu không có NaN, nhưng giá trị thiếu được ghi bằng chuỗi `'unknown'`. Đổi `'unknown'` thành NaN rồi thống kê:

```python
df_na = df.replace('unknown', np.nan)
total = df_na.isnull().sum().sort_values(ascending=False)
percent = (df_na.isnull().sum() / len(df_na) * 100).round(1).sort_values(ascending=False)
missing_data = pd.concat([total, percent], axis=1, keys=['Total', '%'])
missing_data.head(5)
```

**Nhận xét:**

- Dữ liệu không có NaN, nhưng giá trị thiếu được ghi là `'unknown'`.
- Sau khi đổi `'unknown'` thành NaN: `poutcome` thiếu 81,7%, `contact` 28,8%, `education` 4,1%, `job` 0,6%; các thuộc tính còn lại không thiếu.
- `poutcome` thiếu quá nhiều nên nên bỏ; `education`, `job` thiếu ít nên có thể điền bằng mode.
