File không có tiêu đề cột; 3 dòng cuối là các dòng thống kê (min, max, trung bình), không phải dữ liệu đo nên bỏ đi.

```python
df = pd.read_excel(DATA + 'tb_tracking.xlsx', header=None)
# Cuối file có 1 dòng trống và 3 dòng thống kê (min, max, trung bình) -> bỏ
blank = df[df.isnull().all(axis=1)].index
if len(blank):
    df = df.loc[:blank[0] - 1]
df.columns = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']
df['id'] = df['id'].astype(int)
df['device_id'] = df['device_id'].astype(int)
df['time'] = pd.to_datetime(df['time'])
df.head(10)
```

**Nhận xét:**

- File gốc không có tiêu đề, gồm 33.336 dòng đo. Ba dòng cuối (sau một dòng trống) là min, max, trung bình của từng cột nên đã bỏ đi.
- Sau khi đặt tên, các cột `id`, `device_id` là kiểu nguyên, `co_level`, `humidity`, `temperature` là kiểu số thực, `time` là kiểu thời gian.
